### Importaciones


```python
import pandas as pd
import numpy as np
import re
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_val_score

import time
```

### Separación del Dataframe

```python
y=df_train['abandono']
X=df_train.drop(columns=['abandono'])
```

### Clasificación de variables

```python
# Numéricas genéricas
numericas=[
    'nota_acceso', 'creditos_a1', 'superados_a1', 'nota_s1', 'satisfaccion',
    'posts_foro', 'uso_biblioteca', 'eventos', 'tutorias',
    'meses_matriculado', 'talla_zapato', 'horas_trabajo'
]

# Categóricas genéricas
categoricas=[
    'trabaja', 'modalidad', 'comentarios'
]

# Variables que necesitan una transformación específica
fecha =['nacimiento']
moodle = ['horas_moodle']
provincia = ['provincia']
bachillerato = ['bachillerato']
```

### Transformadores específicos

```python
# Transformador para nacimento
class TransformadorNacimiento(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None): return self
    def transform(self, X, y=None):
        X_dt = pd.to_datetime(X.iloc[:, 0], format='%Y-%m-%d', errors='coerce')
        return X_dt.apply(lambda fecha: fecha.year if pd.notna(fecha) else np.nan).to_frame()
    
# Transformador para horas_moodle
class TransformadorHorasMoodle(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None): return self
    def transform(self, X, y=None):
        def limpiar_horas_moodle(valor):
            match = re.search(r'[\d\.]+', str(valor))
            return float(match.group(0)) if match else np.nan
        return X.iloc[:, 0].apply(limpiar_horas_moodle).to_frame()
    
# Transformador para provincia
class TransformadorProvincia(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None): return self
    def transform(self, X, y=None):
        provincias_principales = {'Albacete', 'Cuenca', 'Ciudad Real', 'Toledo'}
        def limpiar_provincia(prov):
            return prov if prov in provincias_principales else 'Otro'
        return X.iloc[:, 0].astype(str).apply(limpiar_provincia).to_frame()
    
# Transformador para bachillerato
class TransformadorBachillerato(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None): return self
    def transform(self, X, y=None):
        diccionario = {
            'científico': 'ciencias', 'ciencias puras': 'ciencias',
            'cientifico-tecnologico': 'tecnologico', 'tecnológico': 'tecnologico',
            'tecnológico industrial': 'tecnologico', 'letras': 'humanidades',
            'económico': 'ciencias sociales', 'ccss': 'ciencias sociales',
            'ciencias': 'ciencias', 'humanidades': 'humanidades',
            'ciencias sociales': 'ciencias sociales', 'artes': 'artes'
        }
        serie = X.iloc[:, 0].astype(str).str.lower().str.strip().replace(diccionario)
        return serie.fillna('missing').to_frame()
```

### Pipelines de Transformación

```python
transformador_numerico = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

transformador_categorico = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

transformador_moodle = Pipeline(steps=[
    ('cleaner', TransformadorHorasMoodle()),
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

transformador_provincia = Pipeline(steps=[
    ('grouper', TransformadorProvincia()),
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

transformador_bachillerato = Pipeline(steps=[
    ('mapper', TransformadorBachillerato()),
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

transformador_nacimiento = Pipeline(steps=[
    ('date_to_year', TransformadorNacimiento()),
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
```

### Column Transformer

```python
preprocesador = ColumnTransformer(
    transformers=[
        ('moodle_pipe', transformador_moodle, moodle),
        ('date_pipe', transformador_nacimiento, fecha),
        ('prov_pipe', transformador_provincia, provincia),
        ('bach_pipe', transformador_bachillerato, bachillerato),
        ('num_pipe', transformador_numerico, numericas),
        ('cat_pipe', transformador_categorico, categoricas),
    ],
    remainder='drop' # Con esta linea, elimino todas las variables que no he transformado (id, nombre, residencia_id, numero_fav, color_fav y grupo_trabajo)
)
```

### Stacking Classifier

Este es el modelo que he elegido implementar e investigar por mi cuenta, ya que me parece muy interesante como funciona. La forma de funcioanmiento de este modeleo, se basa en dos capas: la primera es un conjunto de modelos básicos diversos, como por ejemplo Regresión logistica o un Árbol de decisiones, los cuales son entrenados y generan sus propias prediciones; y la segunda capa es un modelo que se entrena para predecir cual es la predicción final, este modelo no trata directamente con el conjunto de entrenamiento ni nada del dataset, sino que solo 've' las predicciones generadas por la primera capa.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import StackingClassifier

# Modelos que utilizará la capa 1 para generar las predicciones
estimators_base = [
    # 1. Modelo Lineal
    ('log_reg', LogisticRegression(random_state=semilla, solver='liblinear', max_iter=1000)), 
    # 2. Modelo Basado en Árboles
    ('dt', DecisionTreeClassifier(random_state=semilla)), 
]

stacking_model = StackingClassifier(
    estimators=estimators_base, 
    final_estimator=LogisticRegression(random_state=semilla, solver='liblinear'), 
    n_jobs=-1,
    passthrough=False 
)

pipeline = Pipeline([
    ('preprocessor', preprocesador),
    ('classifier', stacking_model)
])

start_time = time.time()
scores = cross_val_score(
    estimator=pipeline,
    X=X,
    y=y,
    cv=cv_strategy,
    scoring='f1_macro',
    n_jobs=-1
)

print('Stacking Classifier\nMedia : ', np.mean(scores), '\nDesviación estandar : ', np.std(scores), '\nTiempo de ejecución :', time.time()-start_time)
```

De estas breves pruebas lo que puedo sacar en claro es que de normal, los modelos ensemble suelen obtener un mejor resultado que los modelos convencionales (especialmente Histagram-Based Gradient Boosting y StackingClassifier), aunque también tardan un tiempo considerablemente superior en ejecutarse.

## 2. Selección de Hiperparámetros

En esta parte, voy a probar los modelos anteriores, variando los hiperparámetros, para ver cuál es la mejor configuración para cada modelo. Además obtendré un muy buen resulyado de score para cada uno de los modelos, por lo que luego me permitirá escoger mejor al indicado. Para ello haré uso de la función facilitada en el cuaderno de ejemplo de esta misma práctica, 'optimize_with_grid_search'.

### Función facilitada

```python
from sklearn.model_selection import GridSearchCV

def optimize_with_grid_search(estimator, param_grid, X_train, y_train, cv_strategy, scoring_metric='accuracy', refit=True):
    n_jobs = -1  # Usar todos los cores de CPU disponibles
    return_train_score = True # Incluir puntuaciones de entrenamiento en los resultados

    grid_search = GridSearchCV(
        estimator=estimator,
        param_grid=param_grid,
        scoring=scoring_metric,
        n_jobs=n_jobs,
        cv=cv_strategy,
        refit=refit, # Pasa refit directamente como se configuró
        return_train_score=return_train_score,
        verbose=1 # Muestra actualizaciones de progreso
    )

    model_name = type(estimator).__name__
    if isinstance(estimator, Pipeline):
        # Obtener el nombre del último paso (clasificador) si es un pipeline
        model_name = type(estimator.steps[-1][1]).__name__

    print(f"--- Iniciando GridSearchCV para {model_name} ---")
    start_time = time.time()
    grid_search.fit(X_train, y_train)
    end_time = time.time()
    print(f"GridSearchCV para {model_name} completado en {end_time - start_time:.2f} segundos.")

    # --- Procesar y Mostrar Resultados ---
    print(f"\nMejores Parámetros encontrados para {model_name}:")
    print(grid_search.best_params_)
    refit_metric_key = refit if isinstance(refit, str) else next(iter(grid_search.scoring)) if isinstance(grid_search.scoring, dict) else 'score'
    print(f"\nMejor Puntuación Media CV ({refit_metric_key}): {grid_search.best_score_:.4f}")

    results_df = pd.DataFrame(grid_search.cv_results_)
    rank_col_name = f"rank_test_{refit_metric_key}"

    # Ordenar por la columna de ranking
    if rank_col_name in results_df.columns:
        results_df = results_df.sort_values(by=rank_col_name)

    # Seleccionar columnas relevantes para mostrar
    param_cols = [col for col in results_df.columns if col.startswith('param_')]
    score_cols = [col for col in results_df.columns if ('mean_test_' in col or 'std_test_' in col or col.startswith('rank_test_'))]
    train_score_cols = [col for col in results_df.columns if ('mean_train_' in col or 'std_train_' in col)]
    time_cols = ['mean_fit_time', 'std_fit_time', 'mean_score_time', 'std_score_time']

    # Asegurar que todas las columnas seleccionadas existen
    cols_to_display = param_cols + [col for col in score_cols if col in results_df.columns] + \
                      [col for col in train_score_cols if col in results_df.columns] + \
                      [col for col in time_cols if col in results_df.columns]
    # Eliminar duplicados si alguna columna aparece en múltiples listas
    cols_to_display = sorted(list(set(cols_to_display)), key=lambda x: (not x.startswith('param_'), x))

    print(f"\nResumen de Resultados GridSearchCV para {model_name}:")
    display(results_df[cols_to_display])
    return grid_search
```

En concreto los valores que he probado son 30 valores diferentes entre 10⁻9 y 1. Cuanto más próximo a 0 es esta varianza más similar será su comportamiento a un modelo Naive Bayes convencional, y cuanto mayor sea esa varianza (más próxima a 1), el modelo se suaviza, y actua en parte como si estuvuera regularizado. Como se puede ver, el mejor valor tanto para train y test, se consigue con un valor de varianza artificial que se encuantra a medio camino, aunque más próximo al 1; lo que nos prodría indicar que el modelo de Naive Bayes para este conjunto de datos, necesita algún tipo de suavizado, aunque no mucho. ### Árbol de decisiones

```python
arbol_pipeline = Pipeline([
    ('preprocessor', preprocesador),
    ('classifier', DecisionTreeClassifier(random_state=semilla)) 
])

arbol_param_grid = {
    'classifier__max_depth': [10, 20, 30, None], # Profundidad máxima (None significa ilimitada)
    'classifier__min_samples_split': [2, 5, 10], # Mínimo de muestras para dividir un nodo
    'classifier__min_samples_leaf': [1, 5, 10], # Mínimo de muestras que debe tener un nodo hoja
    'classifier__criterion': ['gini', 'entropy'] # Función para medir la calidad de la división
}

arbol_grid_search = optimize_with_grid_search(
    estimator=arbol_pipeline,
    param_grid=arbol_param_grid,
    X_train=X,
    y_train=y,
    cv_strategy=cv_strategy,
    scoring_metric='f1_macro'
)

mejor_arbol=arbol_grid_search.best_estimator_
```

En cuanto al árbol de decisiones, el hiperparámetro más relevante parece ser el criterio, ya que como se puede ver, todos los buenos resultados mostrados en la tabla, se han obtenido con 'gini' como criterio y todos los malos con 'entropy'.  
  

Otro que parece muy relevante es la profundidad, ya que se ve como los mejores resultados se han conseguido o sin limite o con limite = 30 (el mayor valor con el que e probado); mientras que los peores resultados se han conseguido con limite = 10 (el mínimo valor que he probado), por lo que la conclusión que se puede sacar es que para este conjunto de datos es mejor un límite de profundidad muy grande o no tener limite directamente.  
  
Algo similar pasa con el mínimo de muestras de cada nodo hoja, aunque al revés, es decir que por lo que se ve en los resultados, los mejores scores se han obtenido con el número mínimo posible para ese hiperparámetro, mientras que los peores se han obtenido con minimo de muestras más grandes, por lo que se puede asumir que para este conjunto de datos es mejor pocas muestras por hoja.  
  
El último hiperparámetro, el número mínimo de muestras que debe de terner un nodo para poder dividirlo, es el que parece más inutil, ya que los 3 mejores scores (y los 3 peores) se han obtenido con la misma convinación de los 3 anteriores hiperparámetros, solo cambiando este. Además se puede ver, especialmente en los mejores scores como un número bajo de este hiperparámetro hace que haya mucho overfittin, llegando el tercer modelo a tener un 1 en el train.

Esto tiene sentido, ya que al fin y al cabo este modelo utiliza árboles de decisiones.  
  
Lo mismo pasa con el mínimo de muestras por hoja, aunque de una maner un poco diferente, ya que aqui si que vemos que hay ocasiones en las que es mejor tener un limite de muestras diferente del mínimo, cosa que en el aŕbol no pasaba.

Si que es cierto que paar árboles iguales (los otros dos hiperparámetros constantes) si que es mejor que hay más modelos, aunque no siempre.

### Adaptive boost

```python
adaptive_boost_pipeline = Pipeline([
    ('preprocessor', preprocesador),
    ('adaboost', AdaBoostClassifier(random_state=semilla, estimator=DecisionTreeClassifier(max_depth=1))) 
])

adaptive_boost_param_grid = {
    'adaboost__n_estimators': [50, 100, 200, 300],
    'adaboost__learning_rate': [0.01, 0.1, 0.5, 1.0], 
}

adaptive_boost_grid_search = optimize_with_grid_search(
    estimator=adaptive_boost_pipeline,
    param_grid=adaptive_boost_param_grid,
    X_train=X,
    y_train=y,
    cv_strategy=cv_strategy,
    scoring_metric='f1_macro'
)

mejor_adaboost=adaptive_boost_grid_search.best_estimator_
```

Se puede apreciar claramente como el hiperparámetro más relevante de este modelo, por lo menos para este conjunto de datos es el ratio de aprendizaje. Realmente lo que hace es que a los modelos bien clasificados se les de más o menos peso en el voto final, pero en la práctica tiene el efecto anteriormente comentado. Hay que tener en cuenta que un valor alto de este parámetro hace que tienda mucho al overfittin.  
  
El segunda parametro, el número de modelos, es mucho menos importante, por lo menos en este caso. Hay que tener en cuenta, que una buena forma de usar este hiperparámetro, es de manera inversamente proporcional al ratio de aprendizaje, ya que es conveniente con un ratio de aprendizaje alto, un número bajo de módelos evitaría el overfittin; y al revés también es conveniente.

### Stacking Classifier

```python
estimators_base = [
    ('log_reg', LogisticRegression(random_state=semilla, solver='liblinear', max_iter=1000)), 
    ('dt', DecisionTreeClassifier(random_state=semilla, max_depth=5)), 
]

tacking_model = StackingClassifier(
    estimators=estimators_base, 
    final_estimator=LogisticRegression(random_state=semilla, solver='liblinear'), 
    n_jobs=-1,
    passthrough=False 
)

stacking_pipeline = Pipeline([
    ('preprocessor', preprocesador),
    ('stacking', stacking_model)
])

stacking_param_grid = {
    # Hiperparámetros para el meta-estimador (final_estimator__C)
    'stacking__final_estimator__C': [0.1, 1.0, 10.0], 
    # Hiperparámetros para el estimador base Regresión Logística (log_reg__C)
    'stacking__log_reg__C': [0.1, 1.0], 
    # Hiperparámetros para el estimador base Árbol de Decisión (dt__max_depth)
    'stacking__dt__max_depth': [3, 7]
}

stacking_grid_search = optimize_with_grid_search(
    estimator=stacking_pipeline,
    param_grid=stacking_param_grid,
    X_train=X, 
    y_train=y, 
    cv_strategy=cv_strategy,
    scoring_metric='f1_macro'
)

mejor_stacking = stacking_grid_search.best_estimator_
```

En este caso, lo primero será explicar un poco del modelo, y es que en este caso, los modelos básicos (de la capa 1) que he escogido son la regresión logistica y el árbol de decisión. El modelo que he elegido como estimador final (capa 2) ha sido también la regresión logística; ya que por lo que he visto es muy común usar este modelo para ambas capas; además de que como lo he implementado como tal, he pensado que estaría bien utilizarlo aqui. Hay un hiperparámetro por cada modelo usado.  
  
Con esto dicho, parece ser que el hiperparámetro más relevante en este caso es el parámetro C de regulación, pero del modelo de regresión logistica de la capa 1, es decir el básico. Esto seguramente se deba a que sea su estimación la más correcta y la que más se tenga en cuenta en la segunda capa.  
  
Lo anterior parece tener más sentido tras comprobar el hiperparámetro del otro modelo básico, la profundidad m-axima del árbol (en este caso solo he elegido usar un hiperparámetro para el árbol para no complicar en exceso el modelo). No parece tener demasiado peso sobre el score, ya que los distintos valores aparecen con los mejores scores y con los peores, por lo que creo que este es el hiperparámetro más irrelevante del modelo en este caso.  
  
Por último, el hiperparámetro del modelo del estiamdor final, que es el mismo que en el anterior modelo de la capa 1; ya que también es una regresión logistica, solo que en este caso le he dado un posible valor más alto, ya que los modelos de la capa 2 tienden menos a sobre ajustar. Por lo que se puede apreciar en los datos de salida, parece que va mejor en valores altos, y como se pede apreciar no esta ligado a un incrementeo significativo del overfitting.

```python
def generar_archivo_csv(pipeline, df, nombre_archivo):

    # Guardo la columna par luego insertarla en el archivo de entrega (ya que lo elimino en la pipeline)
    ids = df['id'].copy() 

    # Predigo la columna 'abandono'
    predicciones_abandono = pipeline.predict(df)
    
    # Construyo el dataframe con los resultados y el id
    df_submission = pd.DataFrame({
        'id': ids,
        'abandono': predicciones_abandono
    })
    
    # Me aseguro de que el tipo de dato generado para la predicción de 'abandono' es int
    df_submission['abandono'] = df_submission['abandono'].astype(int)

    # Genero el csv
    df_submission.to_csv(nombre_archivo, index=False)

    return df_submission
```
